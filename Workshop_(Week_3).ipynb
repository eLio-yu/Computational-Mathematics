(Example 7)
Sine Function Plot

```python
import matplotlib.pyplot as plt
import math

x = []
y = []

start = -4 * math.pi
end = 4 * math.pi
step = 0.1

i = start
while i <= end:
    x.append(i)
    y.append(math.sin(i))
    i += step

plt.plot(x, y, label="sin(x)")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("y = sin(x), -4π ≤ x ≤ 4π")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import math

x = []
y = []

r = 4
center = [2, 2]

theta = 0
step_size = 0.125

while theta <= 2 * math.pi:
    x.append(r * math.cos(theta) + center[0])
    y.append(r * math.sin(theta) + center[1])
    theta += step_size

plt.figure()
plt.plot(center[0], center[1], '.')
plt.plot(x, y)

plt.grid()
plt.xlabel("X")
plt.ylabel("Y")

ax = plt.gca()
ax.set_aspect('equal', adjustable='box')

plt.show()
```

(Example 10)
Parabola Plot

```python
import matplotlib.pyplot as plt

p = 0
q = 0
a = 0.5

step_size = 0.1

y = -5
y_end = 5

x_values = []
y_values = []

while y <= y_end:
    x = 4 * a * (y - p)**2 + q
    x_values.append(x)
    y_values.append(y)
    y += step_size

plt.figure()
plt.title(f"Parabola with vertex at ({p},{q})")

plt.plot(x_values, y_values)
plt.grid()

plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

(Question 10)
Plot a circle with center at (a, b) and radius r. The values of a=b=2, and r=4, using a while loop.

```python
import matplotlib.pyplot as plt
import math

# circle parameters
a = 2  # x-coordinate of center
b = 2  # y-coordinate of center
r = 4  # radius

# initializing lists to store coordinates
x_values = []
y_values = []

# parameter theta from 0 to 2π
theta = 0
theta_end = 2*math.pi
step_size = 0.1

# calculating points of the circle
while theta <= theta_end:
    x = r * math.cos(theta) + a  # x = r*cos(θ) + a
    y = r * math.sin(theta) + b  # y = r*sin(θ) + b
    x_values.append(x)
    y_values.append(y)
    theta += step_size

# plotting the circle
plt.plot(x_values, y_values, '-', color='blue', label='Circle')
plt.plot(a, b, '.', color='red', label='Center')

# adding title, labels, and grid
plt.title('Circle with center at (2,2) and radius 4')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()

# ensuring equal aspect ratio
ax = plt.gca()
ax.set_aspect('equal', adjustable='box')

# displaying the plot
plt.show()
```

(Question 11)
Plot the above circle and the ellipse on the same axes. This is give an idea about the intersection points of the two conic sections.

```python
import matplotlib.pyplot as plt
import math

# CIRLCE PARAMETERS
circle_a = 2  # x-coordinate of center
circle_b = 2  # y-coordinate of center
circle_r = 4  # radius

# initializing lists for circle coordinates
circle_x = []
circle_y = []

theta = 0
theta_end = 2 * math.pi
step_size = 0.1

# calculating points for circle
while theta <= theta_end:
    circle_x.append(circle_r * math.cos(theta) + circle_a)
    circle_y.append(circle_r * math.sin(theta) + circle_b)
    theta += step_size

# ELLIPSE PARAMETERS
ellipse_a = 3  # semi-minor axis
ellipse_b = 5  # semi-major axis
ellipse_center = [2, 2]  # (p,q)

# initializing lists for ellipse coordinates
ellipse_x = []
ellipse_y = []

theta = 0
theta_end = 2 * math.pi

# calculating points for ellipse
while theta <= theta_end:
    ellipse_x.append(ellipse_a * math.cos(theta) + ellipse_center[0])
    ellipse_y.append(ellipse_b * math.sin(theta) + ellipse_center[1])
    theta += step_size

# plotting circle and ellipse on the same axes
plt.plot(circle_x, circle_y, color='blue', label='Circle')
plt.plot(ellipse_x, ellipse_y, color='red', label='Ellipse')

# marking centers
plt.plot(circle_a, circle_b, '.', color='blue')
plt.plot(ellipse_center[0], ellipse_center[1], '.', color='red')

# adding title, labels, grid, legend
plt.title('Circle and Ellipse on the same axes')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()

# ensuring equal aspect ratio
ax = plt.gca()
ax.set_aspect('equal', adjustable='box')

# displaying the plot
plt.show()
```

(Question 12)
Plot a Parabola (𝑦 − 𝑝) = 4𝑎(𝑥 − 𝑞)^2 with vertex at (𝑝,𝑞). The values of p, q, and a can be chosen appropriately.

```python
import matplotlib.pyplot as plt

# parabola parameters
p = 0  # x-coordinate of vertex
q = 0  # y-coordinate of vertex
a = 0.5  # controls width and direction

# defining x range around the vertex
x_start = p - 5
x_end = p + 5
step_size = 0.1

# initializing lists to store coordinates
x_values = []
y_values = []

# calculating points for parabola: (y-q) = 4a(x-p)^2  =>  y = 4a(x-p)^2 + q
x = x_start
while x <= x_end:
    y = 4 * a * (x - p)**2 + q
    x_values.append(x)
    y_values.append(y)
    x += step_size

# plotting the parabola
plt.plot(x_values, y_values, '-', color='green', label='Parabola')

# marking the vertex
plt.plot(p, q, '.', color='red', label='Vertex')

# adding title, labels, grid, legend
plt.title(f'Parabola with vertex at ({p},{q}) and a={a}')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.legend()

# displaying the plot
plt.show()
```
