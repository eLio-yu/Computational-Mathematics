(Example 1)
Solve Linear Systems Graphically

```python
import numpy as np
import matplotlib.pyplot as plt

""" Part 1 - Problem 1: Solve linear systems graphically. """

# generate 100 equidistant points between -5 to 5 for x
x = np.linspace(-5, 5, 100)

# calculate y values for equations
y1 = (5 - x) / 3
y2 = (5 + x) / 2
y3 = 3 - x

# generate figure
fig, ax = plt.subplots()

# plot lines
ax.plot(x, y1, color='r')
ax.plot(x, y2, color='g')
ax.plot(x, y3, color='b')

# add equations
ax.text(0.5, 0.6, '$x+3y = 5$', color='r')
ax.text(-3, 0.5, '$x-2y = -5$', color='g')
ax.text(1.2, 2, '$x+y = 3$', color='b')

# axis limits
ax.set_xlim(-4, 4)
ax.set_ylim(-2, 6)

# draw axes
ax.axvline(color='k', linewidth=1)
ax.axhline(color='k', linewidth=1)

# ticks
ax.set_xticks(list(range(-4, 5)))

ax.set_aspect('equal')
ax.grid(True, ls=':')

plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

(Question: 1)
Rank and Invertibility of Matrix

```python
import numpy as np

A = np.random.randint(1, 10, (5, 5))

print("Matrix A is ")
print(A)

rank = np.linalg.matrix_rank(A)

print("\nRank of Matrix is ")
print(rank)

det = np.linalg.det(A)

print("\nDeterminant is ")
print(det)

if det != 0:
    print("Matrix is invertible")
else:
    print("Matrix is not invertible")
```

Matrix Vector Product and Solve System

```python
import numpy as np

A = np.random.randint(1, 10, (5, 5))

x = np.random.randint(1, 10, 5)

b = np.dot(A, x)

print("Matrix A is ")
print(A)

print("\nVector x is ")
print(x)

print("\nVector b is ")
print(b)

solution = np.linalg.solve(A, b)

print("\nSolution is ")
print(solution)
```

Task 4 (Polynomial Curve Fitting)

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 7])
y = np.array([-7.6, 1, 4.8, 5, 2.8, -0.6, -4, -6.2, -6, 6.4])

# polynomial fitting
coefficients = np.polyfit(x, y, 3)

print("Polynomial Coefficients:")
print(coefficients)

# polynomial equation
p = np.poly1d(coefficients)

# generate smooth curve
x_new = np.linspace(-3, 7, 200)
y_new = p(x_new)

# plot points
plt.scatter(x, y)

# plot polynomial
plt.plot(x_new, y_new)

plt.xlabel("X")
plt.ylabel("Y")

plt.grid(True)

plt.show()
```
