Write a function fib (n) that prompts the number of terms “n” from the user and displays the first “n” terms of the Fibonacci sequence iteratively

```python
def fib(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")
        c = a+b
        a = b
        b = c

n = int(input("Enter the number of terms "))
fib(n)
```

Write a function myPow(x, n) that returns 𝑥𝑛, where and n is a non-negative integer. Do not use the ** operator or the math.pow function. (use one while loop)

```python
def myPow(x, n):
    result = 1
    i = 0

    while i<n:
        result = result*x
        i += 1

    return result

x = int(input("Enter the base (x) "))
n = int(input("Enter the exponent (n) "))

print("Result is", myPow(x, n))
```

Write a function SumDigits(n) that displays the sum of digits of “n”. (hint: SumDigits(234) will be 2+3+4=9)

```python
def SumDigits(n):
    sum = 0
    a = n
    while a>0:
        digit = a%10
        sum = sum+digit
        a = a//10

    print("Sum of digits in", n ,"is", sum)

n = int(input("Enter a number "))
SumDigits(n)
```

Write a function squaredSum that returns the sum of the squares of an array/list of integers such that squaredSum([2,3,4]) gives 29. Write your solution in pure Python to show the logic in your solution; do not use any library functions.

```python
def squaredSum(arr):
    total = 0
    i = 0

    while i<len(arr):
        total = total+(arr[i]*arr[i])
        i += 1

    return total

nums = [5, 6, 7]
print("Sum of squares is", squaredSum(nums))
```

Write a function called sumDigits that is given an integer num and returns the sum of the digits of num. eg, sumDigits(6) gives 21. Do not use any library functions. (if there is one)

```python
def sumDigits(num):
    total = 0
    i = 0

    while i<=num:
        total = total+i
        i += 1

    return total

n = int(input("Enter a number "))
print("Sum is", sumDigits(n))
```
