Question 1
(Write a Python function that takes a list and returns a new list with distinct elements from the first list)

```python
def ele(lst):
    unique_list = []
    for item in lst:
        if item not in unique_list:
            unique_list.append(item)
    return unique_list

sample_list = [1, 2, 3, 3, 3, 3, 4, 5]
print(ele(sample_list))
```

Question 2
(Write a Python function that takes a number as a parameter and checks whether the number is prime or not)

```python
def prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number "))

if prime(num):
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

Question 3
(Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime, then they are known as twin primes)

```python
def prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

print("Twin primes less than 1000 are ")

for i in range(3, 1000, 2):
    if prime(i) and prime(i + 2):
        print(f"({i},{i+2})")
```

Question 4
(A positive integer is called a perfect number if it is equal to the sum of all of its divisors, including 1 but excluding the number itself. For example, 6 = 1 + 2 + 3. Write a function sumOfDivisors that takes a positive integer n and returns the sum of all its divisors (excluding n))

```python
def sumOfDivisors(n):
    total = 0
    for i in range(1, n):
        if n%i == 0:
            total += i
    return total

num = 18
print("Sum of divisors is", sumOfDivisors(num))
```

```python
def sumOfDivisors(n):
    total = 0
    for i in range(1, n):
        if n % i == 0:
            total += i
    return total

num = 7

while True:
    if sumOfDivisors(num) == num:
        print("Next perfect number is", num)
        break
    num += 1
```

Question 6
(Write the python program to find the mean median from the following list:
Marks = [20, 30, 35, 55, 70, 80, 90] Do not use the library function if there is one.)

```python
marks = [20, 30, 35, 55, 70, 80, 90]

total = 0
for i in marks:
    total += i

mean = total/len(marks)

n = len(marks)

if n%2 == 1:
    median = marks[n//2]
else:
    median = (marks[n//2 - 1] + marks[n//2]) / 2

print("Mean is", mean)
print("Median is", median)
```

Question 7
(Write a function that converts a decimal number to binary number)

```python
def bi(n):
    if n == 0:
        return "0"

    binary = ""
    while n > 0:
        remainder = n%2
        binary = str(remainder) + binary
        n = n//2

    return binary

num = 13
print("Binary of", num ,"is", bi(num))
```

Question 8
(Write a function prodDigits() that inputs a number and returns the product of digits of that number)

```python
def prod(n):
    product = 1
    while n>0:
        digit = n%10
        product *= digit
        n = n//10
    return product


num = 157
print("Product of digits of", num ,"is", prod(num))
```

Question 9
(Write a function that will take in an unknown number of arguments and multiply all of them together and run the function for these 2 sets of numbers:

i. 1,2,3,4,5
ii. 12,13,14)

```python
def multiply(*args):
    product = 1
    for num in args:
        product *= num
    return product

resultA = multiply(1, 2, 3, 4, 5)
print("Product of (1,2,3,4,5) is", resultA)

resultB = multiply(12, 13, 14)
print("Product of (12,13,14) is", resultB)
```
