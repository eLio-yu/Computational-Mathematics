Task 2
(multiple pie charts {subplot})

```python
import matplotlib.pyplot as plt

years = [2017, 2018, 2019, 2020, 2021, 2022]
deers =   [18, 25, 33, 106, 137, 44]
elephants = [105, 67, 42, 11, 61, 10]
zebras =  [87, 54, 90, 47, 93, 72]
label = ["Deers", "Elephants", "Zebras"]

for i in range(len(years)):
  x=[]
  x.append(deers[i])
  x.append(elephants[i])
  x.append(zebras[i])
  plt.title(f"Population distribution in {years[i]}")
  plt.pie(x,labels=label)
  plt.show()
```

(Example 4.2) Bar Graph with Grid

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
data = [23, 45, 56, 78, 213]

plt.bar(x, data, width=0.5, color='maroon')

plt.grid(color='#95a5a6', linestyle='--', linewidth=2, axis='y', alpha=0.7)

plt.title("A bar plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

(Question 6)
Test out the following pie chart additional arguments / settings (“...” is for the normal code you have already)

A) myexplode = [0.3, 0, 0, 0]
   plt.pie(... , explode = myexplode)

```python
import matplotlib.pyplot as plt

car_colors = ["Red", "Silver", "Blue", "White", "Green"]
numbers = [123, 230, 78, 193, 12]

myexplode = [0.3, 0, 0, 0, 0]

plt.pie(numbers, labels=car_colors, explode=myexplode)
plt.title("Pie Chart with Explode")
plt.show()
```

B) myexplode = [0.3, 0, 0, 0]
   plt.pie(... , explode = myexplode,shadow = True)

```python
import matplotlib.pyplot as plt

car_colors = ["Red", "Silver", "Blue", "White", "Green"]
numbers = [123, 230, 78, 193, 12]

myexplode = [0.3, 0, 0, 0, 0]

plt.pie(numbers, labels=car_colors, explode=myexplode, shadow=True)
plt.title("Pie Chart with Shadow")
plt.show()
```

C) plt.legend(title = "Car Colours:")

```python
import matplotlib.pyplot as plt

car_colors = ["Red", "Silver", "Blue", "White", "Green"]
numbers = [123, 230, 78, 193, 12]

plt.pie(numbers, labels=car_colors)
plt.legend(title="Car Colours:")
plt.title("Pie Chart with Legend")
plt.show()
```

D) mycolours = ["Magenta", "Green", "Cyan", "Red"]
   plt.pie(... , colors = mycolours)

```python
import matplotlib.pyplot as plt

car_colors = ["Red", "Silver", "Blue", "White", "Green"]
numbers = [123, 230, 78, 193, 12]

mycolours = ["magenta", "green", "cyan", "red", "yellow"]

plt.pie(numbers, labels=car_colors, colors=mycolours)
plt.title("Pie Chart with Custom Colors")
plt.show()
```
