```python
import numpy as np
import matplotlib.pyplot as plt

origin = np.array([0,0])
V = np.array([1,1])
W = np.array([-2.0,-1])
fig, ax = plt.subplots()

ax.quiver(origin[0], origin[1], V[0], V[1],
          angles='xy', scale_units='xy', scale=1, color='r')

ax.quiver(origin[0], origin[1], W[0], W[1],
          angles='xy', scale_units='xy', scale=1, color='b')

ax.text(origin[0]-0.1, origin[1]+0.1, "O")
ax.text(V[0], V[1], "V")
ax.text(W[0]-0.2, W[1]-0.1, "W")
ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_xlabel('X')
ax.set_ylabel('Y')

plt.grid()
plt.show()
```

sub task (a)

```python
import numpy as np
import matplotlib.pyplot as plt

origin = np.array([0,0,0])

U= np.array([1, 0.25 ,0])
V = np.array([0, 1, 0.25])
W = np.array([0.25, 0, 1])

fig = plt.figure()
ax = fig.add_subplot(111,projection='3d')

ax.quiver(origin[0],origin[1],origin[2],
          U[0],U[1],U[2],color='red')

ax.quiver(origin[0],origin[1],origin[2],
          V[0],V[1],V[2],color='blue')

ax.quiver(origin[0],origin[1],origin[2],
          W[0],W[1],W[2],color='yellow')

ax.text(origin[0]-0.1,origin[1],origin[2]-0.2,"O")
ax.text(U[0],U[1],U[2],"U")
ax.text(V[0],V[1],V[2],"V")
ax.text(W[0]-0.2,W[1],W[2],"W")

ax.set_xlim([-1,1])
ax.set_ylim([-1,1])
ax.set_zlim([-1,1])

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.grid()
plt.show()
```

sub task (b)

```python
import numpy as np
import matplotlib.pyplot as plt

origin = np.array([0,0])

U = np.array([-0.5, 0.5])
V = np.array([1,1])
W = U + V

fig, ax = plt.subplots()

ax.quiver(origin[0],origin[1],
          U[0],U[1],
          angles='xy',scale_units='xy',scale=1,color='red')

ax.quiver(origin[0],origin[1],
          V[0],V[1],
          angles='xy',scale_units='xy',scale=1,color='blue')

ax.quiver(origin[0],origin[1],
          W[0],W[1],
          angles='xy',scale_units='xy',scale=1,color='green')

ax.text(origin[0]-0.1,origin[1]+0.1,"O")
ax.text(U[0],U[1],"U")
ax.text(V[0],V[1],"V")
ax.text(W[0],W[1],"W = U+V")

ax.set_xlim([-2,2])
ax.set_ylim([-2,2])

ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.grid()
plt.show()
```

```python
import numpy as np

def magnitude(v):
    return np.sqrt(np.sum(v**2))

U = np.array([1.5,1.5,2.0])
print(magnitude(U))
```

sub task (d)

```python
import numpy as np

def direction_cosines(v):
    mag = np.sqrt(np.sum(v**2))
    return v/mag

O = np.array([1.5,1.5,2.0])
print(direction_cosines(O))
```

```python
import numpy as np

def dot_product(a,b):
    return np.dot(a,b)

a = np.array([1.5,1.5,2.0])
b = np.array([-1.5,0.5,2.0])

print(dot_product(a,b))
```

sub task (f)

```python
import numpy as np

def cross_product(a,b):
    return np.cross(a,b)

a = np.array([1.5,1.5,2.0])
b = np.array([-1.5,0.5,2.0])

print(cross_product(a,b))
```
